```python
# Import library Pandas untuk manipulasi dan analisis data
import pandas as pd

# Import library NumPy untuk operasi numerik dan array
import numpy as np

print('Library berhasil diimpor')
print(f'   Pandas  versi: {pd.__version__}')
print(f'   NumPy   versi: {np.__version__}')
```

```python
# Load dataset Titanic
url = 'https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv'

df = pd.read_csv(url)

print('Dataset berhasil dimuat')
df.head()
```

## Q1. Berapa jumlah baris dan kolom pada dataset Titanic? Kolom apa saja yang tersedia?

```python
# df.shape mengembalikan tuple (baris, kolom)
jumlah_baris, jumlah_kolom = df.shape
print(f'Jumlah Baris : {jumlah_baris}')
print(f'Jumlah Kolom : {jumlah_kolom}')

# Menampilkan nama kolom beserta tipe datanya
print('\nDaftar Kolom yang Tersedia:')
for i, col in enumerate(df.columns, 1):
    # enumerate mulai dari 1 agar nomor urut lebih rapi
    print(f'  {i:2}. {col} ({df[col].dtype})')
```

## Q2. Kolom mana saja yang memiliki missing values? Berapa jumlah dan proporsinya (%) per kolom?

```python
# isnull() menghasilkan True/False per cell, sum() menjumlahkan True-nya
missing_count = df.isnull().sum()

# Proporsi = jumlah missing dibagi total baris, dikali 100
missing_pct = (df.isnull().sum() / len(df)) * 100
missing_df = pd.DataFrame({
    'Jumlah Missing': missing_count,
    'Proporsi (%)': missing_pct.round(2)
})

# Filter hanya kolom yang benar-benar punya missing values
missing_df = missing_df[missing_df['Jumlah Missing'] > 0].sort_values('Proporsi (%)', ascending=False)

print('Kolom dengan Missing Values:')
print(missing_df)
print(f'\nTotal kolom dengan missing: {len(missing_df)} dari {df.shape[1]} kolom')
```

## Q3. Berapa persen penumpang yang selamat (Survived=1)? Gunakan operasi NumPy atau Pandas.

```python
# Konversi kolom Survived ke NumPy array
survived_array = df['Survived'].values

# np.mean() pada array 0/1 langsung menghasilkan proporsi yang selamat
pct_selamat = np.mean(survived_array) * 100
pct_tidak = 100 - pct_selamat

# np.sum() menghitung jumlah penumpang yang selamat (nilai 1)
print(f'Total penumpang      : {len(survived_array)}')
print(f'Selamat (Survived=1) : {np.sum(survived_array == 1)} ({pct_selamat:.2f}%)')
print(f'Tidak selamat (=0)   : {np.sum(survived_array == 0)} ({pct_tidak:.2f}%)')
```

## Q4. Filter penumpang wanita dari kelas 1 (Pclass=1). Berapa jumlahnya dan berapa ratarata usianya?

```python
# Kondisi 1: kolom Sex harus 'female'
# Kondisi 2: kolom Pclass harus 1
# Operator & menggabungkan dua kondisi (keduanya harus True)
wanita_kelas1 = df[(df['Sex'] == 'female') & (df['Pclass'] == 1)]

# Menampilkan jumlah hasil filter
print(f'Jumlah wanita kelas 1 : {len(wanita_kelas1)} orang')

# .mean() otomatis mengabaikan NaN saat menghitung rata-rata
print(f'Rata-rata usia        : {wanita_kelas1["Age"].mean():.2f} tahun')
```

## Q5. Gunakan groupby untuk menghitung tingkat keselamatan (mean Survived) per kelas penumpang. Kelas mana yang paling tinggi?

```python
# groupby('Pclass') mengelompokkan data berdasarkan kelas tiket (1, 2, 3)
# .mean() pada kolom Survived (0/1) menghasilkan proporsi penumpang yang selamat
print(df.groupby('Pclass')['Survived'].mean())
```

## Q6. Buat analisis tambahan berdasarkan variabel lain yang Anda anggap menarik. Sertakan interpretasi singkat.
Pada analisis ini saya memilih 3 variabel tambahan yang menarik:
1. **Jenis Kelamin (Sex)** — apakah gender mempengaruhi peluang selamat?
2. **Kelas x Gender** — bagaimana kombinasi kelas dan gender mempengaruhi keselamatan?
3. **Kelompok Usia (AgeGroup)** — apakah usia muda lebih diprioritaskan saat evakuasi?

```python

print('=== Survival Rate by Gender ===')
print(df.groupby('Sex')['Survived'].mean().mul(100).round(2).to_string())


print('\n=== Survival Rate (%) — Kelas x Gender ===')
pivot = df.groupby(['Pclass', 'Sex'])['Survived'].mean().mul(100).round(2).unstack()
pivot.index = ['Kelas 1', 'Kelas 2', 'Kelas 3']
print(pivot.to_string())


df['AgeGroup'] = pd.cut(df['Age'], bins=[0,12,18,35,60,100],
    labels=['Anak', 'Remaja', 'Dewasa Muda', 'Dewasa', 'Lansia'])
print('\n=== Survival Rate (%) — Kelompok Usia ===')
print(df.groupby('AgeGroup', observed=True)['Survived'].mean().mul(100).round(2).to_string())
```

A. Survival Rate by Gender:
Wanita memiliki tingkat keselamatan jauh lebih tinggi (sekitar 74%) dibanding pria (sekitar 19%).
Hal ini mencerminkan kebijakan evakuasi "women and children first" yang diterapkan awak kapal.

B. Kelas x Gender:
Wanita di semua kelas memiliki survival rate lebih tinggi dari pria di kelas yang sama.
Ini menunjukkan bahwa gender lebih dominan daripada kelas dalam menentukan peluang selamat.

C. Kelompok Usia:
Anak-anak (0–12 tahun) memiliki survival rate tertinggi, mendukung prioritas evakuasi untuk anak-anak.
Kelompok lansia justru memiliki survival rate paling rendah.
